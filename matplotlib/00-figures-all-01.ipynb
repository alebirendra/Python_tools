```python
# '/home/birendra/Desktop/apjp/a66/v24/e15/all/avg'

import matplotlib.pyplot as plt
import numpy as np
from matplotlib import rcParams
from matplotlib.ticker import LogLocator
import pandas as pd
from sklearn.metrics import r2_score

def safe_r2_score(y_true, y_pred):
    # Handle cases where R² might be negative or undefined
    try:
        return max(0, r2_score(y_true, y_pred))
    except:
        return 0




from scipy.optimize import curve_fit
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import rcParams



# Configure matplotlib to use LaTeX with Times New Roman
plt.rcParams.update({
    "text.usetex": True,
    'pdf.fonttype': 42,
    "font.family": "serif",
    "font.serif": ["Times New Roman"],
    'font.size': 10,
    'axes.labelsize': 10,
    'axes.titlesize': 8,
    'legend.fontsize': 10,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
    'lines.linewidth': 2.0,
    'axes.linewidth': 0.7,
    'xtick.major.width': 0.7,
    'ytick.major.width': 0.7,
    'xtick.major.size': 2.5,
    'ytick.major.size': 2.5,
    'xtick.minor.size': 1.5,
    'ytick.minor.size': 1.5,
    'figure.dpi': 800,
    'figure.constrained_layout.use': True,
    'figure.constrained_layout.h_pad': 0.1,
    'figure.constrained_layout.w_pad': 0.1,
    'legend.frameon': True,
    'legend.edgecolor': 'black',
    'legend.fancybox': False,
    
    # Tick parameters
    'xtick.top': True,
    'xtick.bottom': True,
    'ytick.left': True,
    'ytick.right': True,
    'xtick.direction': 'in',
    'ytick.direction': 'in',
    'xtick.minor.width': 0.7,  
    'ytick.minor.width': 0.7,
})


colors = [
    '#006BA4', '#FF800E', '#ABABAB', '#595959', '#5F9ED1',
    '#C85200', '#898989', '#A2C8EC', '#FFBC79', '#CFCFCF'
]

# Line and marker styles
line_styles = ['-', '--', ':', '-.']
markers = ['o', 's', '^', 'D', 'v', 'p', 'h', '*', 'x', '+']
label = ['Calculated data', 'Exponential fit' ]



def logistic_model_00(t, C, r ):

    return C - (C - 1) * np.exp(-t*r)


initial_guess = [1, 0.0035]  # 

# Initialize storage for results
results1 = []

x = []
all_y = []
all_yerr = []

# Initialize storage for results
list_pe = [0.0, 34.0, 50.0, 100.0, 200.0, 600.0] 
list_pe = [ 0, 50, 70, 100, 200, 600 ] 

for i in list_pe:
    y_curves=[]
    filename = 'pe%s.0' %i
    pe = i

   
    df_avg = pd.read_csv('/home/birendra/Desktop/apjp/a66/v24/e15/all/avg/trans/%s/sample40_avg_TransClustStats.csv' %(filename) )
    df_avg0n = df_avg.iloc[0].copy()
    df_avg0n['gT'] = 0
    df_avg0 = pd.concat([pd.DataFrame([df_avg0n]), df_avg], ignore_index=True)

    t_data = df_avg0['gT'] #...  # Load time data for p
    N_data = df_avg0['NC']  #...  # Load N data for p
    N_data_err = df_avg0['NCstd']  #...  # Load N data for p
    y_curves.append(N_data.values)
    all_yerr.append(N_data_err.values)  
    x = t_data.values
    try:


        params, covariance = curve_fit(logistic_model_00, t_data, N_data, p0=initial_guess, bounds=([0, 0], 
                             [np.inf, np.inf]), maxfev=10000)
        K_fit, r_fit = params
        
                
        t_fit = np.linspace(1, 20001, 1000)
        #t_fit = t_data
        N_fit = logistic_model_00(t_fit, K_fit, r_fit ) # K_fit * (1 -  np.exp(-r_fit * t_fit))
        N_fit2 = logistic_model_00(t_data, K_fit, r_fit ) # K_fit * (1 -  np.exp(-r_fit * t_fit))

        
        residuals = N_data - logistic_model_00(t_data, K_fit, r_fit )

        mae = np.mean(np.abs(residuals))
       # print(f"MAE: {mae:.4f}")
        rmse = np.sqrt(np.mean(residuals**2))
       # print(f"RMSE: {rmse:.4f}")
       # ss_res = np.sum(residuals**2)
       # ss_tot = np.sum((N_data - np.mean(N_data))**2)
       # r_squared = 1 - (ss_res / ss_tot)
       # print(f"R²: {r_squared:.4f}")
        r2sc = safe_r2_score(N_data, N_fit2 ) # f"{r_squared:.3f}"
        r2score = f"{r2sc:.2f}"
        C = np.round(K_fit, 2)

        #b = f'{
        #results5.append({'Pe': pe, 'beta': beta, 'r': rate, 't0':t0b, 'mae':mae, 'rmse':rmse, 'R2score': r2score})
        results1.append({'Pe': pe, 'C': C, 'r': r_fit, 'mae':mae, 'rmse':rmse, 'R2score': r2score})

        y_curves.append(N_fit2)
    

    except RuntimeError:
        print(f"Failed to fit for p = {pe}")

    all_y.append(y_curves)
 


df = pd.DataFrame(results1)
df['R2score'] = pd.to_numeric(df['R2score'])
r2mean = np.round(df['R2score'].mean(), 2)



# Create a single figure
fig, ax = plt.subplots(figsize=(3.5, 3.1))

ew = 0.7
ct = 0.7
lw = 1.5

# Plot all data and fits on the same axes
for i, (y_curve, yerr, result) in enumerate(zip(all_y, all_yerr, results1)):
    # Adjust error bars to not go below 1
    adjusted_yerr = np.copy(yerr)
    lower_bounds = y_curve[0] - yerr
    mask = lower_bounds < 1
    adjusted_yerr[mask] = y_curve[0][mask] - 1  # Limit error to not go below 1
    
    # Plot data points with error bars
    ax.plot(x, y_curve[0], color=colors[i], linestyle=line_styles[0], lw=lw,
           marker=markers[i], markevery=2, markersize=5, alpha=0.8, 
           label=f"Pe = {result['Pe']} ")
    
    ax.errorbar(x[::2], y_curve[0][::2], adjusted_yerr[::2], linestyle='', lw=lw, capsize=1.75,
               color=colors[i], markevery=2, markersize=5, elinewidth=ew, capthick=ct, alpha=0.8)
    
    # Plot fitted curve
    ax.plot(x, y_curve[1], color=colors[i], linestyle=line_styles[1], lw=lw,
           marker='', alpha=0.8)

ax.text(0.4, 11., 'R$^2$ = %s' %r2mean)

# Configure the single plot
ax.set_xscale('symlog')
ax.set_xlim(-0.5, 32000)
ax.set_ylim(-1., 28)
ax.minorticks_on()
ax.xaxis.set_minor_locator(LogLocator(base=10.0, subs='auto', numticks=10))
ax.grid(False)
ax.tick_params(axis="both", which='both', top=True, bottom=True, 
              left=True, right=True, direction='in')

# Set labels
ax.set_xlabel(r"$t\mathrm{/\tau_D}$", labelpad=3)
ax.set_ylabel(r"$N_\mathrm{C}$", labelpad=3)

# Add horizontal line at y=1 to indicate the minimum value
ax.axhline(y=0, color='gray', linestyle=':', alpha=0.7, linewidth=1)

# Add legend
leg = ax.legend(ncol=1, frameon=True, framealpha=1,
               edgecolor='black', fancybox=False, loc='upper left')
leg.get_frame().set_linewidth(0.5)

# Add label
ax.text(0.05, 0.95, '(a)', transform=ax.transAxes,
       fontsize=12, fontweight='bold', va='top')

plt.tight_layout()

# Save figures
plt.savefig('./fig_final/v24/e15/01_aniso_exponential_fit_combined_v3.png', bbox_inches="tight", pad_inches=0.02, format='png')
plt.savefig('./fig_final/v24/e15/01_aniso_exponential_fit_combined_v3.pdf', bbox_inches="tight", pad_inches=0.02, format='pdf')
plt.savefig('./fig_final/v24/e15/01_aniso_exponential_fit_combined_v3.svg', bbox_inches="tight", pad_inches=0.02, format='svg')
```
